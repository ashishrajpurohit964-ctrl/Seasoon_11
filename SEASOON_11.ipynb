1.Use plt.subplots() to create a 2x2 grid of subplots and plot four different types of charts (line, bar, scatter, and pie) using any sample data of your choice.

```python
import matplotlib.pyplot as plt

# Sample data
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [20, 35, 30, 40, 50]

products = ["Pizza", "Burger", "Biryani", "Pasta"]
orders = [40, 30, 50, 25]

hours = [1, 2, 3, 4, 5]
marks = [45, 50, 60, 70, 85]

categories = ["Food", "Travel", "Shopping", "Entertainment"]
expenses = [4000, 2000, 3000, 1000]

fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# 1. Line chart
axes[0, 0].plot(days, sales, marker="o")
axes[0, 0].set_title("Daily Sales")
axes[0, 0].set_xlabel("Day")
axes[0, 0].set_ylabel("Sales")

# 2. Bar chart
axes[0, 1].bar(products, orders, color="brown")
axes[0, 1].set_title("Orders by Product")
axes[0, 1].set_xlabel("Product")
axes[0, 1].set_ylabel("Orders")

# 3. Scatter plot
axes[1, 0].scatter(hours, marks, color="brown")
axes[1, 0].set_title("Study Hours vs Marks")
axes[1, 0].set_xlabel("Study Hours")
axes[1, 0].set_ylabel("Marks")

# 4. Pie chart
axes[1, 1].pie(expenses, labels=categories, autopct="%1.1f%%")
axes[1, 1].set_title("Monthly Expenses")

plt.tight_layout()

plt.show()
```

2.Plot a comparison of average delivery times for Zomato, Swiggy, and Domino's using a bar chart, and style each bar with a different color, linestyle, and linewidth using Matplotlib plot styling options

```python

platforms = ["Zomato", "Swiggy", "Domino's"]

delivery_time = [35, 30, 25]

plt.bar(
    platforms,
    delivery_time,
    color=["pink", "brown", "gold"],
    edgecolor="black",
    linewidth=2
)

plt.title("Average Delivery Time")

plt.xlabel("Food Delivery Platform")

plt.ylabel("Delivery Time (Minutes)")

plt.show()
```

3.Create a multi-axis chart that shows the number of Instagram followers (left y-axis) and average daily posts (right y-axis) for five influencers. Use different colors and markers for each axis.


```python

influencers = ["Influencer 1 :- virat kohli ", "Influencer 2 :- shraddha kapoor", "Influencer 3 :- Bhuvan Bam",
               "Influencer 4 :- Ashish", "Influencer 5 :- Disha patani"]

followers = [100000, 250000, 180000, 400000, 320000]

daily_posts = [2, 4, 3, 5, 4]

fig, ax1 = plt.subplots()

ax1.plot(
    influencers,
    followers,
    color="brown",
    marker="o",
    label="Followers"
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Instagram Followers")
ax1.set_title("Instagram Followers and Average Daily Posts")

ax2 = ax1.twinx()

ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s",
    label="Daily Posts"
)

ax2.set_ylabel("Average Daily Posts")
plt.show()
```

4.Plot the number of tickets sold for five recent Bollywood movies (categorical) versus their IMDB ratings (numeric) using a scatter plot. Add annotations to display the movie names above each point

```python

movies = ["Drishyam", "Drishyam 2", "Drishyam (conclusion)", "Power", "Spy"]

tickets = [185000, 120000, 95000, 70000, 60000]

ratings = [8.2, 5.8, 6.9, 6.1, 6.0]

plt.scatter(tickets, ratings)

for i in range(len(movies)):
    plt.annotate(movies[i], (tickets[i], ratings[i]),
                 xytext=(5, 5), textcoords="offset points")

plt.xlabel("Number of Tickets Sold")
plt.ylabel("IMDb Rating")
plt.title("Tickets Sold vs IMDb Ratings of Bollywood Movies")

plt.show()
```

5.Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales point with the label 'Big Billion Days'.<br><br><em><strong>Hint:</strong> Use the ax.annotate() function to place the label at the correct data point.</em>

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", 
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [120, 135, 150, 140, 160, 175, 
         190, 200, 180, 450, 210, 230]

fig, ax = plt.subplots()

ax.plot(months, sales, marker="o")

ax.annotate("Big Billion Days",
            xy=("Oct", 450),
            xytext=("Aug", 350),
            arrowprops=dict(arrowstyle="->"))

ax.set_xlabel("Month")
ax.set_ylabel("Sales (₹ Crore)")
ax.set_title("Flipkart Monthly Sales")

plt.show()
```
